# 13 Capstone

本章目标：组合前十二章知识，实现服务产品、研发和测试协作的生产级 Agent。

## 项目范围

1. 读取需求、技术方案、测试计划和项目记录
2. 提取目标、范围、验收标准、风险和待确认问题
3. 查询任务、缺陷、测试结果和发布状态
4. 生成任务拆分、测试建议、风险清单和发布摘要
5. 在写入系统、发送通知和修改状态前请求人工确认

## 项目模块

1. 后端 API
2. 知识库 ingestion
3. Agent service
4. tool calling
5. human review UI 或 CLI
6. tracing
7. evals
8. audit log
